```python
from sklearn.model_selection import train_test_split #Split the dataset into training and testing sets
from sklearn.preprocessing import StandardScaler #Scalling Data
from sklearn.linear_model import LogisticRegression #Modelling Side
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve #Model Evaluation
from sklearn.pipeline import Pipeline #Train-test split, scaling, and logistic regression in a single pipeline
from sklearn.compose import ColumnTransformer #Data Transformation
from sklearn.preprocessing import OneHotEncoder #Data Transformation
```

```python
df.head()
```

```python
df_model = df.drop(columns=["nameOrig", "nameDest", "isFlaggedFraud"])
```

```python
df_model.head()
```

```python
categorical_features = ["type"]
numerical_features = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest", "newbalanceDest", "balanceDiffOrig", "balanceDiffDest"]
```

```python
y = df_model["isFraud"]
X = df_model.drop(columns=["isFraud"])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("numerical", StandardScaler(), numerical_features),
        ("categorical_features", OneHotEncoder(), categorical_features)
    ],
    remainder="passthrough"
)
```

```python
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("clf", LogisticRegression(class_weight="balanced", max_iter=1000))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
confusion = confusion_matrix(y_test, y_pred)
confusion
```

```python
pipeline.score(X_test, y_test) *100
```

```python
import joblib
joblib.dump(pipeline, "fraud_detection_model.pkl") 
```
